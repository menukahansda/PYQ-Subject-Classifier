## Load the text extracted

```python
with open(TARGET_DIR / "notebooks" / "texts.pkl", "rb") as f:
    texts = pickle.load(f)
```

# SBERT Encode

```python

model = SentenceTransformer('all-MiniLM-L6-v2')
embeddings = model.encode(texts)
```

# Clustering

```python
db = DBSCAN(eps=0.8, min_samples=3, metric='cosine').fit(embeddings)
labels = db.labels_
print(labels)
```
